```python
event_counts = pd.crosstab(
    [ad_events["user_id"], ad_events["ad_id"]],
    ad_events["event_type"]
).reset_index()

display(event_counts.head())
```

```python
event_counts = event_counts.rename(columns={
    "impression": "Impressions",
    "click": "Clicks",
    "purchase": "Purchases"
})
```

```python
df = df.merge(
    campaigns,
    on="campaign_id",
    how="left"
)
```

```python
df = df.merge(
    users,
    on="user_id",
    how="left"
)
```

```python
df["CTR"] = np.where(
    df["Impression"] > 0,
    df["Click"] / df["Impression"],
    0
)
```

```python
df["Conversion_Rate"] = np.where(
    df["Click"] > 0,
    df["Purchase"] / df["Click"],
    0
)
```

```python
df["Purchase_Rate"] = np.where(
    df["Impression"] > 0,
    df["Purchase"] / df["Impression"],
    0
)
```

```python
df["CTR_Percent"] = df["CTR"] * 100
df["Conversion_Percent"] = df["Conversion_Rate"] * 100
df["Purchase_Rate_Percent"] = df["Purchase_Rate"] * 100
```

```python
group_summary = df.groupby("ad_type").agg(
    Users=("user_id", "nunique"),
    Impressions=("Impression", "sum"),
    Clicks=("Click", "sum"),
    Purchases=("Purchase", "sum")
)

display(group_summary)
```

```python
group_summary["CTR"] = (
    group_summary["Clicks"] /
    group_summary["Impressions"] *
    100
)
```

```python
group_summary["Conversion_Rate"] = (
    group_summary["Purchases"] /
    group_summary["Clicks"] *
    100
)
```

```python
group_summary["Purchase_Rate"] = (
    group_summary["Purchases"] /
    group_summary["Impressions"] *
    100
)
```

```python
from statsmodels.stats.proportion import proportions_ztest

# Get data for the baseline ad type (e.g., 'Carousel')
baseline_ad_type = 'Carousel'
baseline_clicks = group_summary.loc[baseline_ad_type, 'Clicks']
baseline_impressions = group_summary.loc[baseline_ad_type, 'Impressions']

print(f"Comparing CTRs against {baseline_ad_type} (Baseline):\n")

# Iterate through other ad types and perform z-test
for ad_type in group_summary.index:
    if ad_type == baseline_ad_type:
        continue

    other_clicks = group_summary.loc[ad_type, 'Clicks']
    other_impressions = group_summary.loc[ad_type, 'Impressions']

    # Perform proportions z-test for two independent samples
    count = np.array([baseline_clicks, other_clicks])
    nobs = np.array([baseline_impressions, other_impressions])

    z_stat, p_value = proportions_ztest(
        count=count,
        nobs=nobs,
        alternative='two-sided' # or 'smaller', 'larger'
    )

    print(f"  {ad_type} vs {baseline_ad_type}:")
    print(f"    Z-statistic: {z_stat:.4f}")
    print(f"    P-value: {p_value:.4f}")
    if p_value < 0.05:
        print("    There is a statistically significant difference in CTRs.")
    else:
        print("    There is no statistically significant difference in CTRs.")
    print("\n")
```

```python
alpha = 0.05

if p_value < alpha:
    print("CTR difference is statistically significant.")
else:
    print("CTR difference is not statistically significant.")
```

```python
from statsmodels.stats.proportion import proportions_ztest

# Get data for the baseline ad type (e.g., 'Carousel')
baseline_ad_type = 'Carousel'
baseline_purchases = group_summary.loc[baseline_ad_type, 'Purchases']
baseline_clicks = group_summary.loc[baseline_ad_type, 'Clicks']

print(f"Comparing Conversion Rates against {baseline_ad_type} (Baseline):\n")

# Iterate through other ad types and perform z-test
for ad_type in group_summary.index:
    if ad_type == baseline_ad_type:
        continue

    other_purchases = group_summary.loc[ad_type, 'Purchases']
    other_clicks = group_summary.loc[ad_type, 'Clicks']

    # Perform proportions z-test for two independent samples
    count = np.array([baseline_purchases, other_purchases])
    nobs = np.array([baseline_clicks, other_clicks])

    z_stat_conv, p_value_conv = proportions_ztest(
        count=count,
        nobs=nobs,
        alternative='two-sided'
    )

    print(f"  {ad_type} vs {baseline_ad_type}:")
    print(f"    Z-statistic: {z_stat_conv:.4f}")
    print(f"    P-value: {p_value_conv:.4f}")
    if p_value_conv < 0.05:
        print("    There is a statistically significant difference in Conversion Rates.")
    else:
        print("    There is no statistically significant difference in Conversion Rates.")
    print("\n")
```

```python
if p_value_conv < 0.05:
    print("Conversion difference is statistically significant.")
else:
    print("Conversion difference is not statistically significant.")
```

```python
from statsmodels.stats.proportion import proportion_confint

for version in group_summary.index:

    clicks_value = group_summary.loc[version, "Clicks"]
    impressions_value = group_summary.loc[version, "Impressions"]

    lower, upper = proportion_confint(
        clicks_value,
        impressions_value,
        alpha=0.05,
        method="wilson"
    )

    print(
        version,
        "CTR 95% CI:",
        lower * 100,
        "-",
        upper * 100
    )
```

```python
for version in group_summary.index:

    purchase_value = group_summary.loc[version, "Purchases"]
    click_value = group_summary.loc[version, "Clicks"]

    lower, upper = proportion_confint(
        purchase_value,
        click_value,
        alpha=0.05,
        method="wilson"
    )

    print(
        version,
        "Conversion 95% CI:",
        lower * 100,
        "-",
        upper * 100
    )
```

```python
df["Purchase_Flag"] = np.where(
    df["Purchase"] > 0,
    1,
    0
)
```

```python
print(df.columns.tolist())
```

```python
model_df = df[
    [
        "ad_type", # Corrected from 'Ad_Version'
        "Impression", # Corrected from 'Impressions'
        "Click", # Corrected from 'Clicks'
        "Purchase_Flag"
    ]
].copy()

model_df = pd.get_dummies(
    model_df,
    columns=["ad_type"], # Corrected from 'Ad_Version'
    drop_first=True
)

display(model_df.head())
```

```python
from sklearn.model_selection import train_test_split

X = model_df.drop(
    columns=["Purchase_Flag"]
)

y = model_df["Purchase_Flag"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

model.fit(
    X_train,
    y_train
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

y_pred = model.predict(X_test)

print("Accuracy:",
      accuracy_score(y_test, y_pred))

print("Precision:",
      precision_score(y_test, y_pred, zero_division=0))

print("Recall:",
      recall_score(y_test, y_pred, zero_division=0))

print("F1 Score:",
      f1_score(y_test, y_pred, zero_division=0))
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred
)

plt.title("Decision Tree Confusion Matrix")
plt.show()
```

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(20,10))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=["No Purchase", "Purchase"],
    filled=True,
    rounded=True
)

plt.show()
```
